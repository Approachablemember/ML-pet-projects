Step 1. Import libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline

# Settings
sns.set(style="whitegrid")
```

Step: 2. Load the dataset

```python
import seaborn as sns
df = sns.load_dataset('titanic')

df.head()
```

Step 3. Dataset overview

```python
print("Dataset shape:", df.shape)

df.info()

df.describe(include='all')
```

Step 4. Missing values check

```python
df.isnull().sum().sort_values(ascending=False)
```

Step 5. Univariate Analysis

```python
sns.countplot(x='sex', data=df)
plt.title('Gender Distribution')
plt.show()

sns.histplot(df['age'].dropna(), kde=True)
plt.title('Age Distribution')
plt.show()
```

Step 6. Bivariate Analysis

```python
sns.countplot(x='survived', hue='sex', data=df)
plt.title('Count Survival by Gender')
plt.show()

sns.countplot(x='survived', hue='age', data=df)
plt.title('Age distribution by survival')
plt.show()
```

Step 7. Correlation Heatmap

```python
corr = df.corr(numeric_only=True)

plt.figure(figsize=(10,6))
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show
```

Step 8. Summarization

Based on received results we can see that:
- Woman have higher survive rate than man
- Children and Teenagers have higher survival rate that adult and old people
- Alone Adult male have a bit higher survival rate
- People with higher Fare have a bit higher survival rate
